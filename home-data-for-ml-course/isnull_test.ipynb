```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df_train = pd.read_csv('train.csv')
# 目的変数であるSalePriceを抽出
df_train_y = df_train['SalePrice']
# 目的変数を削除
df_train = df_train.drop('SalePrice', axis=1)
```

```python
df_test = pd.read_csv('test.csv')
```

```python
# 欠損率の計算
missing_ratio_train = df_train.isnull().sum() / len(df_train)

# グラフの大きさを指定
plt.figure(figsize=(16, 6))

# 欠損率のグラフを描画
missing_ratio_train.plot(kind='bar')
```

```python
# 欠損率の計算
missing_ratio_test = df_test.isnull().sum() / len(df_test)

# グラフの大きさを指定
plt.figure(figsize=(16, 6))

# 欠損率のグラフを描画
missing_ratio_test.plot(kind='bar')
```

```python
# missing_ratioが0以上の列を削除(df_train)
df_train = df_train.drop(missing_ratio_train[missing_ratio_train > 0].index, axis=1)

# missing_ratioが0以上の列を削除(df_test)
df_test = df_test.drop(missing_ratio_test[missing_ratio_test > 0].index, axis=1)
```

```python
# 欠損値がある列の列名を取得(df_train)
missing_columns_train = df_train.columns[df_train.isnull().sum() > 0]

# 欠損値がある列の列名を取得(df_test)
missing_columns_test = df_test.columns[df_test.isnull().sum() > 0]
```

```python
# 欠損値がある列データの中でも、カテゴリカルデータのものだけを抽出(df_train)
missing_categorical_columns_train = df_train[missing_columns_train].select_dtypes(include='object').columns

# 欠損値がある列データの中でも、カテゴリカルデータのものだけを抽出(df_test)
missing_categorical_columns_test = df_test[missing_columns_test].select_dtypes(include='object').columns
```

```python
# 欠損値がある列データの中でも、カテゴリカルデータのものだけを削除(df_train)
df_train = df_train.drop(missing_categorical_columns_train, axis=1)

# 欠損値がある列データの中でも、カテゴリカルデータのものだけを削除(df_test)
df_test = df_test.drop(missing_categorical_columns_test, axis=1)
```
